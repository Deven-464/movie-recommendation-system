```python
# evaluates an expression node or a string containing a Python literal or container display
import ast
data['cast'] = data['cast'].apply(ast.literal_eval)
data['genres'] = data['genres'].apply(ast.literal_eval)
data['crew'] = data['crew'].apply(ast.literal_eval)

```

```python
def make_genres_list(x):
    gen = []
    st = ' '
    for i in x:
        if i.get('name') == "Science Fication":
            sci = "Sci-Fi"
            gen.append(sci)
        else: 
            gen.append(i.get('name'))
    if gen == []:
        return np.nan
    else:
        return (st.join(gen))   

 
```

```python
#data['genres_list'] = data['genres'].map(lambda x: make_genres_list(x))
data['genres_list'] = data['genres'].map(lambda x: make_genres_list(x))
```

```python
data['genres_list']
```

```python
def make_actors1_list(x):
    casts = []
    for i in x:
        casts.append(i.get('name'))
    if casts == []:
        return np.nan
    else:
        return casts[0]
```

```python
data['actor_1_name'] = data['cast'].map(lambda x: make_actors1_list(x))
```

```python
data['actor_1_name']
```

```python
def make_actors2_list(x):
    casts = []
    for i in x:
        casts.append(i.get('name'))
    if casts == [] or len(casts)<=1:
        return np.nan
    else:
        return casts[1]
```

```python
data['actor_2_name'] = data['cast'].map(lambda x: make_actors2_list(x))
```

```python
data['actor_2_name']
```

```python
def make_actors3_list(x):
    casts = []
    for i in x:
        casts.append(i.get('name'))
    if casts == [] or len(casts)<=2:
        return np.nan
    else:
        return (casts[2])
```

```python
import numpy as np

def make_actors3_list(x):
    casts = [i.get('name') for i in x]
    # if fewer than 3 actors, there's no third actor (index 2)
    if len(casts) < 3:
        return np.nan
    return casts[2]
```

```python
data['actor_3_name'] = data['cast'].map(lambda x: make_actors3_list(x))
```

```python
data['actor_3_name']
```

```python
def get_directors(x):
    dt = []
    st = " "
    for i in x:
        if i.get('job') == 'Director':
            dt.append(i.get('name'))
    if dt == []:
        return np.nan
    else:
        return (st.join(dt))
```

```python
data['director_name'] = data['crew'].map(lambda x: get_directors(x))
```
